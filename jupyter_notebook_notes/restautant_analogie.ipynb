### Restaurant-/Küchen-Analogie (Architekturverständnis)

Zur besseren Verständlichkeit der Architektur wurde im Projekt eine Restaurant-/Küchen-Analogie eingeführt.
Diese Analogie dient ausschließlich der didaktischen Unterstützung und verdeutlicht die Rollen und Verantwortlichkeiten der einzelnen Komponenten, ohne die technische Architektur zu beeinflussen.

#### Zuordnung der Komponenten zur Restaurant-/Küchen-Analogie

| Komponente | Rolle in der Analogie | Beschreibung |
|-----------|----------------------|--------------|
| app.py | 🚪 Restaurant-Eingang | Startet die Anwendung, initialisiert die Infrastruktur über `setup.py`, wählt den Servierstil (UI-Backend) und stößt das Rendering des Dashboards an. |
| setup.py | 👷‍♂️ Bau-Team der Küche | Baut die komplette Service-Kette auf und verdrahtet Datenbank, Repositories, WorkflowService, ProgressService und ViewModelBuilder zu einer einsatzbereiten Anwendung. |
| dashboard_page_streamlit.py | 🧑‍💼🍽️ Restaurant-Manager | Orchestriert den Ablauf der Streamlit-Seite, empfängt den Nutzer und stellt sicher, dass die richtigen UI-Komponenten im passenden Servierstil gerendert werden. |
| UI-Komponenten | 🥗💁‍♂️ Dekorateur | Nehmen fertige ViewModels entgegen und visualisieren diese optisch ansprechend (Layout, Diagramme, Tabellen), ohne Geschäftslogik zu enthalten. |
| UI-Adapter | ❇️ Servirstil | legt den Stil des Restaurants fest bzw. wie die Gerichte präsentiert werden z.B. Essen im Dunkeln (CLI, keine visuelle Darstellung), Streamlit (offene Showküche, da direkte Rückmeldung) |
| core/__init__.py | 📋 Küchenplan | Koordiniert die Initialisierung von Repositories, ConnectionProvider, WorkflowService und ProgressService und definiert deren Zusammenspiel. |
| ViewModelBuilder | 💁‍♂️🍽️ Portionierer | Übersetzt DTOs in UI-spezifische ViewModels, ergänzt Präsentationslogik und stellt sicher, dass die UI ausschließlich mit ViewModels arbeitet. |
| ViewModels | 🍽️ Teller | Enthalten die vollständig vorbereiteten Daten für die UI. Die Benutzeroberfläche visualisiert diese ohne weitere Logik. |
| ProgressService | 🧑‍🍳 Koch | Bereitet Rohdaten aus dem WorkflowService auf, aggregiert und analysiert sie und stellt die Ergebnisse als DTOs bereit. |
| DTOs | ❄️📦 Kühlschrank-Boxen | Transportobjekte für vorverarbeitete, UI-nahe Daten. Dienen als stabile Schnittstelle zwischen Logikschicht und Präsentationsschicht. |
| WorkflowService | ↔️ Küchen-Lager-Koordination | Koordiniert fachliche Abläufe, setzt Domänenregeln durch und vermittelt zwischen Repositories und höheren Services. |
| Models | 🥦 Zutat / Lebensmittel | Repräsentieren fachliche Kernobjekte der Domäne (z. B. Studiengang, Kurs). Enthalten nur Zustand und einfache Domänenlogik, keine Auswertungen oder Darstellung. |
| db/repositories/__init__.py | 📦📋 Lagerkatalog | Bündelt alle Repository-Interfaces und deren Implementierungen an einer zentralen Stelle und stellt sie der Anwendung gesammelt zur Verfügung. |
| Repositories | 📦 Lagerverwaltung | Legt fest, welche Aktionen auf Domänenobjekten möglich sind (z. B. finden, anlegen, löschen), ohne die konkrete Datenbank zu kennen. |
| SQLiteRepositories | 📦💁‍♂️ Regalmanager | Konkrete Implementierung eines Repository-Interfaces. Führt Datenbankoperationen für ein Domänenobjekt mittels SQLite aus. |
| ConnectionProvider | 👨‍💼🗝️ Lagermeister | Abstrakte Schnittstelle zur Bereitstellung von Datenbankverbindungen. Kapselt den Zugriff auf die Datenbank und erlaubt den Austausch der konkreten Implementierung. |
| SQLiteConnectionProvider | 🗝️ Lagerschlüssel (SQLite) | Öffnet und schließt Datenbankverbindungen zur SQLite-Datenbank und stellt sicher, dass Verbindungen korrekt erzeugt und freigegeben werden. |
| DBConnection | 🚪 Lagerzugang / Tür | Definiert das Interface für eine konkrete Datenbankverbindung (öffnen, ausführen, schließen), ohne Wissen über Domäne oder Persistenzlogik. |






Das Lager entspricht der Datenhaltung, in der Zutaten in Form von persistierten Domänenobjekten gespeichert sind. Die Repositories übernehmen die Rolle der Lagerverwaltung und kapseln den Zugriff auf die Datenbank. Der ConnectionProvider fungiert dabei als Lagermeister, der den Zugang zum Lager organisiert und Datenbankverbindungen bereitstellt. Die DBConnection entspricht der Lagertür, über die der tatsächliche Zugriff auf das Lager erfolgt. Konkrete SQLite-Repositories übernehmen die Rolle einzelner Regalmanager, die gezielt mit bestimmten Zutaten arbeiten.

Der WorkflowService koordiniert fachliche Abläufe und stellt sicher, dass Domänenregeln eingehalten werden, vergleichbar mit der Organisation von Abläufen zwischen Lager und Küche. Er entscheidet, welche Zutaten wann benötigt werden und wie sie verarbeitet werden dürfen.

Der ProgressService fungiert als Koch, der Rohdaten aus dem WorkflowService aufbereitet, aggregiert und für konkrete Auswertungen vorbereitet. Die Ergebnisse werden in Form von DTOs (Kühlschrank-Boxen) bereitgestellt, die bereits auf die Bedürfnisse einzelner Dashboard-Ansichten zugeschnitten sind. Der ViewModelBuilder übernimmt anschließend die Rolle des „Anrichters“, indem er diese vorbereiteten Daten in UI-spezifische ViewModels überführt und für die Darstellung aufbereitet (Anrichten auf Tellern).

Die Benutzeroberfläche entspricht dem Gastraum des Restaurants. Sie interagiert ausschließlich mit den ViewModels (Tellern) und enthält weder Geschäftslogik noch direkte Datenzugriffe. Die UI-Komponenten übernehmen dabei die Rolle der Dekoration und Präsentation der Gerichte. Der UI-Adapter definiert den Servierstil und ermöglicht den Austausch der konkreten UI-Implementierung (z. B. Streamlit oder alternative Darstellungsformen), ohne Änderungen an der Anwendungslogik vorzunehmen.
