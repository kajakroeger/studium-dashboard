#### Architekturübersicht – Kurzbeschreibung

Die Anwendung ist in vier Schichten gegliedert: Präsentationsschicht (UI), Logikschicht, Domänenschicht sowie Datenzugriffsschicht.  

Die **UI-Schicht** umfasst die Streamlit-Oberfläche, den UI-Adapter sowie den ViewModelBuilder und die ViewModels. Der ViewModelBuilder erzeugt UI-spezifische ViewModels auf Basis vorverarbeiteter DTOs, die vom ProgressService bereitgestellt werden. Der UI-Adapter dient als Schnittstelle zwischen UI und Anwendung und kapselt die konkrete Implementierung der Benutzeroberfläche (derzeit Streamlit). Dadurch bleibt die Präsentationsschicht austauschbar und offen für alternative Darstellungsformen (z. B. CLI oder Web-UI). Die UI-Schicht enthält keine Geschäftslogik und greift nicht direkt auf die Datenbank zu. 

Die **Logikschicht** besteht aus dem WorkflowService und dem ProgressService. Der WorkflowService koordiniert fachliche Abläufe sowie Domänenregeln und arbeitet ausschließlich mit den Repository-Interfaces. Der ProgressService greift auf den WorkflowService zurück, um Analyse- und Auswertungslogik bereitzustellen. Die Ergebnisse werden über DTOs an die Präsentationsschicht übergeben.  

Die **Domänenschicht** wird durch die Models (z. B. Student, Studiengang, Einschreibung, Kurs, Bearbeitung, Prüfung) repräsentiert. 
Diese Modelle bilden die fachlichen Kernobjekte der Anwendung ab und enthalten Domänenzustand sowie einfache Domänenlogik. 
Sie werden vom WorkflowService verarbeitet und von den Repositories persistiert, sind jedoch unabhängig von UI-, DTO- oder Datenbankdetails.

Die **Datenzugriffsschicht** kapselt den Zugriff auf die Datenbank über Repository-Interfaces und konkrete SQLite-Implementierungen. Die SQLite-Implementierungen nutzen einen ConnectionProvider, um DB-Verbindungen bereitzustellen. Dadurch bleiben Datenbankdetails austauschbar, ohne die höheren Schichten zu beeinflussen.
