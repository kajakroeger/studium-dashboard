### Instanzierung 

def `__init__()` ist der Konstruktor. Er wird automatisch ausgeführt, wenn ein neues Student-Objekt erzeugt wird. <br>
self ist die Referenz auf das neue Objekt <br>

```python
# student.py

class Student:
    # Konstruktor
    def __init__(self, name: str, email: str, matrikel_nr: str):
        # Instanzattribute (Zustand des Objekts)
        self.name = name
        self.email = email
        self.matrikel_nr = matrikel_nr

    def zeige_info(self) -> dict:
        """Gibt fachliche Infos des Objekts zurück."""
        return {
            "name": self.name,
            "email": self.email,
            "matrikel_nr": self.matrikel_nr,
        }
```

Die Methode zeige_info() könnte dann in der app.py aufgerufen werden, um die gespeicherten Daten des Studierenden anzuzeigen

```python
# app.py

def main():
    # Objekt instanziieren
    student = Student("Ash Ketchum", "ash@uni.de", "M1234567")
    
    # Methode ausführen – z.B. für eine einfache Ausgabe
    print(student.zeige_info())

if __name__ == "__main__":
    main()
```

Da die UI mit Streamlit geplant ist, würde anstelle von print(), st.write(student.zeige_info()) verwendet werden

### Sichtbarkeit

In Python gibt es keinen echten Zugriffsschutz, sondern nur Konventionen:

| Sichtbarkeit | Schreibweise           | Bedeutung                                                                      |
|--------------|------------------------|--------------------------------------------------------------------------------|
| public       | kein Unterstrich       | frei für alle                                                                  |
| protected    | _geschützt             | nur innerhalb der Klasse oder Unterklassen (Konvention, nicht erzwungen)       |
| private      | __privat               | Name Mangling: wird zu _Klassenname__privat, erschwert versehentlichen Zugriff |


```python
# pruefung.py

class Pruefung:
    def __init__(self, note: float, versuch_nr: int):
        self.note = note                    # public    -> von außen zugreifbar
        self._versuch_nr = versuch_nr       # protected -> Konvention: nur intern/Vererbung
        self.__bestanden = note <= 4.0      # private   -> Name Mangling

    def zeige_info(self):
        return f"Note: {self.note}, Versuch: {self._versuch_nr}, Bestanden: {self.__bestanden}"
```

```python
# Demonstration der Sichtbarkeit 

pr = Pruefung(2.3, 1)

# regulärer Zugriff auf Attribute: über Methode
print("zeige_info():", pr.zeige_info())

# direkter Zugriff auf ein public Attribut 
print("Direkter Zugriff auf public note:", pr.note)

# direkter Zugriff auf eine protected Attribut (möglich, aber nicht empfohlen)
print("Zugriff auf _versuch_nr (protected, nur Konvention):", pr._versuch_nr)

# direkter Zugriff auf ein  private Attribut (führt zu Fehler)
try:
    print(pr.__bestanden)
except AttributeError as e:
    print("Direkter Zugriff auf __bestanden -> Fehler:", e)

# direkter Zugriff auf eon private Attribut über Name Mangling
print("Zugriff über Name Mangling:", pr._Pruefung__bestanden)


```

### Instanzmethode, Klassenmethode & statische Methode

**Klassenattribute** wie _anzahl werden für alle Instanzen gemeinsam gespeichert. Im Beispielfall ist es protected und damit nur für interne Zwecke gedacht. 
<br></br>
**Instanzmethoden** wie zeige_info() sind einem konkreten Objekt Klasse zugeordnet. 
<br></br>
**Klassenmethoden** wie get_anzahl(cls) greifen nicht auf ein einzelnes Objekt zu, sondern auf die gesamte Klasse. 
<br></br>
**Statische Methoden** wie ist_gueltige_matrikelnr() sind Hilfsfunktionen, die zur Klasse gehören aber keine Objekt- oder Klassenattribute benötigen.

```python
class Student:
    _anzahl = 0   # Klassenattribut

    def __init__(self, name: str, matrikelnummer: str):
        self.name = name
        self.matrikelnummer = matrikelnummer
        Student._anzahl += 1

    # Instanzmethode
    def zeige_info(self):
        """Gibt Informationen zu dieser Instanz zurück"""
        return f"Name: {self.name}, Matrikelnummer: {self.matrikelnummer}"

    # Klassenmethode
    @classmethod
    def get_anzahl(cls):
        """Liefert die aktuelle Anzahl der Student-Objekte zurück"""
        return cls._anzahl

    # Statische Methode
    @staticmethod
    def ist_gueltige_matrikelnummer(mnr: str) -> bool:
        """Hilfsfunktion: prüft, ob eine Matrikelnummer gültig ist"""
        return mnr.isdigit() and len(mnr) == 7


# Student-Objekte instanziieren
s1 = Student("Ash Ketchum", "1234567")
s2 = Student("Misty Waterflower", "9876543")

# Aufruf Instanzmethode
print(s1.zeige_info())

# Aufruf Klassenmethode
print("Anzahl Studierende:", Student.get_anzahl())

# Aufruf statische Methode
print("Prüfung Matrikelnummer:", Student.ist_gueltige_matrikelnummer("1234567"))
print("Prüfung Matrikelnummer:", Student.ist_gueltige_matrikelnummer("X999"))
```

### magische Methoden

Python stellt Methoden bereit, die automatisch ausgeführt werden können, z.B. `__str__` zur Textausgabe eines Objektes. <br>

```python
class Student:
    def __init__(self, name: str, matrikelnummer: str):
        self.name = name
        self.matrikelnummer = matrikelnummer

    def __str__(self):
        return f"Student: {self.name}, Matrikelnummer: {self.matrikelnummer}"

# Student-Objekt instanziieren
s1 = Student("Ash Ketchum", "M1234567")
# Student-Objekt ausgeben
print(s1)   #  __str__() wird automatisch aufgerufen
```

### @property-Dekorator

@property macht aus einer Methode ein Property, das sich wie ein Attribut verhält, aber im Hintergrund eine Methode aufruft. 
<br></br>
Mit @...setter (z.B. @note.setter) kann eine Setter-Methode an das Property gehängt und optional eine Schreiblogik (z.B. Validierung des Wertes, bevor es gesetzt wird) hinzugefügt werden. <br>
Der eigentliche Wert der Note wird in einem „geschützten“ Attribut (_note) gespeichert. Damit bleibt der Zugriff nach außen einfach (obj.note), während intern die Kontrolle über Lese- und Schreibvorgänge erhalten bleibt.

```python
class Pruefung:
    def __init__(self, note: float, versuch_nr: int):
        self._note = None                 
        self.note = note                  # ruft den Setter auf, validiert den Wert und speichert ihn in _note 
        self._versuch_nr = versuch_nr     

    @property
    def note(self) -> float:
        """Getter für note"""
        return self._note

    @note.setter
    def note(self, value: float):
        """Setter für note mit Validierung"""
        if not (1.0 <= value <= 6.0):
            raise ValueError("Die Note muss zwischen 1.0 und 6.0 liegen")
        self._note = value

    def zeige_info(self) -> str:
        """Beispiel für fachliche Ausgabe"""
        return f"Versuch {self._versuch_nr}, Note: {self._note}"


# Test
p1 = Pruefung(2.3, 1)
print(p1.zeige_info())

p1.note = 3.7
print("Neue Note:", p1.note)

try:
    p1.note = 7.0
except ValueError as e:
    print("Fehler:", e)
```
